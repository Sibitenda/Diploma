## Vector OperationsCommon vector operations include:- Addition- Subtraction- Dot product- Cross product (3D)

```python
# Define vectors
v1 = np.array([2, 3])
v2 = np.array([4, 1])

# Vector addition
addition = v1 + v2

# Vector subtraction
subtraction = v1 - v2

# Dot product
# v1​⋅v2​=v11​v21​+v12​v22​+⋯+v1n​v2n​
dot_product = np.dot(v1, v2)

print("Addition:", addition)
print("Subtraction:", subtraction)
print("Dot Product:", dot_product)
```

```python
# Cross product example (3D)

# Define two 3D vectors
v3 = np.array([1, 0, 0])
v4 = np.array([0, 1, 0])

# Compute the cross product
cross_product = np.cross(v3, v4)

# Display the result
print("Cross Product:", cross_product)
```

## Parametric Line RepresentationA line can be represented using vectors:P(t) = P0 + tV

```python
# Starting point
P0 = np.array([1, 1])

# Direction vector
V = np.array([2, 1])

# Parameter
t = 2

# Compute a point on the line
P = P0 + t * V

# Display the result
print("Point on line:", P)
```

## 2D Rotation

```python
# 2D Rotation using Homogeneous Coordinates

import numpy as np

# Define the rotation angle (in degrees)
theta = np.radians(45)

# Define the rotation matrix
R = np.array([
    [np.cos(theta), -np.sin(theta), 0],
    [np.sin(theta),  np.cos(theta), 0],
    [0,              0,             1]
])

# Define the point to rotate (homogeneous coordinates)
point = np.array([2, 3, 1])

# Apply the rotation
rotated_point = R @ point

# Display the results
print("Rotation Matrix R:")
print(R)

print("Original Point:", point)
print("Rotated Point:", rotated_point)
```

```python
# 2D Rotation using Homogeneous Coordinates with Visualization

import numpy as np
import matplotlib.pyplot as plt

# Define the rotation angle (in degrees)
theta = np.radians(45)

# Define the rotation matrix
R = np.array([
    [np.cos(theta), -np.sin(theta), 0],
    [np.sin(theta),  np.cos(theta), 0],
    [0,              0,             1]
])

# Define the point to rotate (homogeneous coordinates)
point = np.array([2, 3, 1])

# Apply the rotation
rotated_point = R @ point

# Extract x and y coordinates
x_original, y_original = point[0], point[1]
x_rotated, y_rotated = rotated_point[0], rotated_point[1]

# Plot the points
plt.figure()

# Original point
plt.scatter(x_original, y_original)
plt.text(x_original, y_original, " Original (2,3)")

# Rotated point
plt.scatter(x_rotated, y_rotated)
plt.text(x_rotated, y_rotated, " Rotated")

# Draw lines from origin
plt.plot([0, x_original], [0, y_original])
plt.plot([0, x_rotated], [0, y_rotated])

# Axis settings
plt.axhline(0)
plt.axvline(0)
plt.grid(True)
plt.title("2D Rotation of a Point (45°)")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```

## 2D Reflection (across X-axis)

```python
# 2D Reflection across the X-axis with visualization

import numpy as np
import matplotlib.pyplot as plt

# Define the point in homogeneous coordinates
point = np.array([2, 3, 1])

# Define the reflection matrix across the X-axis
reflection_x = np.array([
    [1, 0, 0],
    [0, -1, 0],
    [0, 0, 1]
])

# Apply the reflection
reflected_point = reflection_x @ point

# Extract coordinates
x_original, y_original = point[0], point[1]
x_reflected, y_reflected = reflected_point[0], reflected_point[1]

# Display numerical results
print("Original Point:", point)
print("Reflected Point:", reflected_point)

# Plot the points
plt.figure()

# Original point
plt.scatter(x_original, y_original)
plt.text(x_original, y_original, " Original")

# Reflected point
plt.scatter(x_reflected, y_reflected)
plt.text(x_reflected, y_reflected, " Reflected")

# Draw lines from origin
plt.plot([0, x_original], [0, y_original])
plt.plot([0, x_reflected], [0, y_reflected])

# Axes and grid
plt.axhline(0)
plt.axvline(0)
plt.grid(True)

plt.title("Reflection Across the X-axis")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```

```python
# Reflection of a Triangle Across the X-axis

import numpy as np
import matplotlib.pyplot as plt

# Define triangle vertices in homogeneous coordinates
triangle = np.array([
    [1, 1, 1],
    [3, 1, 1],
    [2, 3, 1]
])

# Reflection matrix across the X-axis
reflection_x = np.array([
    [1, 0, 0],
    [0, -1, 0],
    [0, 0, 1]
])

# Apply reflection to each vertex
reflected_triangle = (reflection_x @ triangle.T).T

# Extract coordinates
x = triangle[:,0]
y = triangle[:,1]

xr = reflected_triangle[:,0]
yr = reflected_triangle[:,1]

# Plot original triangle
plt.figure()

plt.plot(
    np.append(x, x[0]),
    np.append(y, y[0]),
    marker='o',
    label="Original Triangle"
)

# Plot reflected triangle
plt.plot(
    np.append(xr, xr[0]),
    np.append(yr, yr[0]),
    marker='o',
    label="Reflected Triangle"
)

# Draw axes
plt.axhline(0)
plt.axvline(0)

plt.grid(True)
plt.legend()

plt.title("Reflection of a Triangle Across the X-axis")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```

## 2D Shearing

```python
# 2D Shearing Transformation with Visualization

import numpy as np
import matplotlib.pyplot as plt

# Define triangle vertices (homogeneous coordinates)
triangle = np.array([
    [1, 1, 1],
    [3, 1, 1],
    [2, 3, 1]
])

# Define the shear matrix (shear along the x-direction)
shear = np.array([
    [1, 1, 0],
    [0, 1, 0],
    [0, 0, 1]
])

# Apply shear transformation
sheared_triangle = (shear @ triangle.T).T

# Extract coordinates
x = triangle[:,0]
y = triangle[:,1]

xs = sheared_triangle[:,0]
ys = sheared_triangle[:,1]

# Plot original triangle
plt.figure()

plt.plot(
    np.append(x, x[0]),
    np.append(y, y[0]),
    marker='o',
    label="Original Triangle"
)

# Plot sheared triangle
plt.plot(
    np.append(xs, xs[0]),
    np.append(ys, ys[0]),
    marker='o',
    label="Sheared Triangle"
)

# Draw axes
plt.axhline(0)
plt.axvline(0)

plt.grid(True)
plt.legend()

plt.title("2D Shearing Transformation")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```

## Combining Multiple Transformations

```python
# Combined Transformations Visualization
# Scaling -> Rotation -> Translation

import numpy as np
import matplotlib.pyplot as plt

# Original point (homogeneous coordinates)
point = np.array([2, 3, 1])

# Scaling matrix
S = np.array([
    [2, 0, 0],
    [0, 2, 0],
    [0, 0, 1]
])

# Rotation matrix (45 degrees)
theta = np.radians(45)

R = np.array([
    [np.cos(theta), -np.sin(theta), 0],
    [np.sin(theta),  np.cos(theta), 0],
    [0,              0,             1]
])

# Translation matrix
T = np.array([
    [1, 0, 2],
    [0, 1, 1],
    [0, 0, 1]
])

# Combine transformations
combined_matrix = T @ R @ S

# Apply transformation
result = combined_matrix @ point

print("Result after combined transformation:", result)

# Extract coordinates
x_original, y_original = point[0], point[1]
x_new, y_new = result[0], result[1]

# Plot points
plt.figure()

# Original point
plt.scatter(x_original, y_original)
plt.text(x_original, y_original, " Original")

# Transformed point
plt.scatter(x_new, y_new)
plt.text(x_new, y_new, " Transformed")

# Draw vectors from origin
plt.plot([0, x_original], [0, y_original])
plt.plot([0, x_new], [0, y_new])

# Axis settings
plt.axhline(0)
plt.axvline(0)
plt.grid(True)

plt.title("Combined Transformations (Scale → Rotate → Translate)")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```

# Week 4: 3D Transformations

```python
# 3D Translation using Homogeneous Coordinates

import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Define a 3D point in homogeneous coordinates
point3d = np.array([2, 3, 4, 1])

# Define the 3D translation matrix
T3 = np.array([
    [1, 0, 0, 1],  # translate x by +1
    [0, 1, 0, 2],  # translate y by +2
    [0, 0, 1, 3],  # translate z by +3
    [0, 0, 0, 1]
])

# Apply the translation
translated3d = T3 @ point3d

print("Original Point:", point3d)
print("3D Translated Point:", translated3d)

# Extract coordinates
x, y, z = point3d[:3]
xt, yt, zt = translated3d[:3]

# Plot the points
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Original point
ax.scatter(x, y, z)
ax.text(x, y, z, "Original")

# Translated point
ax.scatter(xt, yt, zt)
ax.text(xt, yt, zt, "Translated")

# Draw a line showing the movement
ax.plot([x, xt], [y, yt], [z, zt])

# Labels
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

ax.set_title("3D Translation Transformation")

plt.show()
```

```python
# 3D Translation Example with Labeled Vertices

import numpy as np
import matplotlib.pyplot as plt

# Define triangle vertices (homogeneous coordinates)
triangle3d = np.array([
    [1, 1, 1, 1],   # A
    [3, 1, 2, 1],   # B
    [2, 3, 1, 1]    # C
])

labels = ["A", "B", "C"]

# Translation matrix
T3 = np.array([
    [1, 0, 0, 2],   # move x by +2
    [0, 1, 0, 1],   # move y by +1
    [0, 0, 1, 3],   # move z by +3
    [0, 0, 0, 1]
])

# Apply translation
translated_triangle = (T3 @ triangle3d.T).T

# Extract coordinates
x = triangle3d[:,0]
y = triangle3d[:,1]
z = triangle3d[:,2]

xt = translated_triangle[:,0]
yt = translated_triangle[:,1]
zt = translated_triangle[:,2]

# Plot
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Plot original triangle
ax.plot(np.append(x,x[0]), np.append(y,y[0]), np.append(z,z[0]), marker='o', label="Original")

# Plot translated triangle
ax.plot(np.append(xt,xt[0]), np.append(yt,yt[0]), np.append(zt,zt[0]), marker='o', label="Translated")

# Label original vertices
for i, label in enumerate(labels):
    ax.text(x[i], y[i], z[i], label)

# Label translated vertices
for i, label in enumerate(labels):
    ax.text(xt[i], yt[i], zt[i], label + "'")

# Draw arrows showing movement
for i in range(3):
    ax.plot([x[i], xt[i]], [y[i], yt[i]], [z[i], zt[i]], linestyle="dashed")

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

ax.set_title("3D Translation with Vertex Labels")
ax.legend()

plt.show()
```

## 3D Scaling

```python
# 3D Scaling Example with Connected Shapes

import numpy as np
import matplotlib.pyplot as plt

# Define original points (homogeneous coordinates)
points = np.array([
    [1, 1, 1, 1],   # A
    [2, 1, 1, 1],   # B
    [2, 2, 1, 1],   # C
    [1, 2, 1, 1]    # D
])

labels = ["A", "B", "C", "D"]

# Scaling matrix
S3 = np.array([
    [2, 0, 0, 0],
    [0, 2, 0, 0],
    [0, 0, 2, 0],
    [0, 0, 0, 1]
])

# Apply scaling
scaled_points = (S3 @ points.T).T

# Extract coordinates
x = points[:,0]
y = points[:,1]
z = points[:,2]

xs = scaled_points[:,0]
ys = scaled_points[:,1]
zs = scaled_points[:,2]

# Create plot
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Plot original shape
ax.plot(np.append(x, x[0]),
        np.append(y, y[0]),
        np.append(z, z[0]),
        marker='o')

# Plot scaled shape
ax.plot(np.append(xs, xs[0]),
        np.append(ys, ys[0]),
        np.append(zs, zs[0]),
        marker='o')

# Label original points
for i, label in enumerate(labels):
    ax.text(x[i], y[i], z[i], label)

# Label scaled points
for i, label in enumerate(labels):
    ax.text(xs[i], ys[i], zs[i], label + "'")

# Draw transformation lines
for i in range(len(points)):
    ax.plot([x[i], xs[i]],
            [y[i], ys[i]],
            [z[i], zs[i]],
            linestyle="dashed")

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

ax.set_title("3D Scaling with Connected Shapes")

plt.show()
```

# Projection## Orthographic Projection

```python
# Clear Orthographic Projection Visualization

import numpy as np
import matplotlib.pyplot as plt

# Define 3D square
square3d = np.array([
    [1, 1, 3],  # A
    [4, 1, 3],  # B
    [4, 4, 3],  # C
    [1, 4, 3]   # D
])

labels = ["A","B","C","D"]

# Orthographic projection matrix
P = np.array([
    [1,0,0],
    [0,1,0]
])

# Compute projection
proj2d = (P @ square3d.T).T

# Convert projected points back to 3D plane z=0
proj3d = np.column_stack((proj2d, np.zeros(len(proj2d))))

# Extract coordinates
x,y,z = square3d.T
xp,yp,zp = proj3d.T

# Plot
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Original square
ax.plot(np.append(x,x[0]), np.append(y,y[0]), np.append(z,z[0]),
        color="blue", marker="o", label="Original 3D Shape")

# Projected square
ax.plot(np.append(xp,xp[0]), np.append(yp,yp[0]), np.append(zp,zp[0]),
        color="red", marker="o", label="Projected Shape")

# Projection lines
for i in range(len(square3d)):
    ax.plot([x[i],xp[i]],[y[i],yp[i]],[z[i],zp[i]],
            linestyle="dashed", color="gray")

# Labels
for i,label in enumerate(labels):
    ax.text(x[i],y[i],z[i],label)
    ax.text(xp[i],yp[i],zp[i],label+"'")

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

ax.set_title("Orthographic Projection onto the XY Plane")
ax.legend()

plt.show()
```

## Perspective Projection (simple example)

```python
# Perspective Projection with Clear Visualization

import numpy as np
import matplotlib.pyplot as plt

# Camera distance (projection plane at z = 0)
d = 5

# Define a 3D square
square3d = np.array([
    [1, 1, 2],   # A
    [4, 1, 2],   # B
    [4, 4, 4],   # C
    [1, 4, 4]    # D
])

labels = ["A", "B", "C", "D"]

# Perspective projection
projected_points = []

for x, y, z in square3d:
    xp = d * x / (d + z)
    yp = d * y / (d + z)
    projected_points.append([xp, yp, 0])   # projection plane z=0

projected_points = np.array(projected_points)

# Extract coordinates
x, y, z = square3d.T
xp, yp, zp = projected_points.T

# Create 3D figure
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Plot original 3D square
ax.plot(np.append(x,x[0]),
        np.append(y,y[0]),
        np.append(z,z[0]),
        marker='o',
        label="Original 3D Shape")

# Plot projected square
ax.plot(np.append(xp,xp[0]),
        np.append(yp,yp[0]),
        np.append(zp,zp[0]),
        marker='o',
        label="Projected Shape")

# Draw projection rays
for i in range(len(square3d)):
    ax.plot([0, x[i]],
            [0, y[i]],
            [0, z[i]],
            linestyle="dotted")

    ax.plot([x[i], xp[i]],
            [y[i], yp[i]],
            [z[i], zp[i]],
            linestyle="dashed")

# Label points
for i, label in enumerate(labels):
    ax.text(x[i], y[i], z[i], label)
    ax.text(xp[i], yp[i], zp[i], label+"'")

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

ax.set_title("Perspective Projection")

ax.legend()
plt.show()
```

# Applying Transformations to Real Data (Iris Dataset)We treat the first three features as **3D coordinates**.

```python
# Load a real dataset (Iris dataset)

from sklearn.datasets import load_iris
import numpy as np

# Load dataset
iris = load_iris()

# Extract the first three features to treat them as 3D coordinates
X = iris.data[:, :3]

# Display dataset information
print("Dataset shape:", X.shape)
print()

# Show the first five 3D points
print("First 5 points (treated as 3D coordinates):")
print(X[:5])
```

```python
# Rotate the dataset around the Z-axis

import numpy as np

# Define rotation angle
theta = np.radians(45)

# Rotation matrix about the Z-axis
Rz = np.array([
    [np.cos(theta), -np.sin(theta), 0],
    [np.sin(theta),  np.cos(theta), 0],
    [0,              0,             1]
])

# Apply rotation to all points in the dataset
rotated_data = X @ Rz.T

# Display results
print("Rotation Matrix (Rz):")
print(Rz)
print()

print("First 5 original points:")
print(X[:5])
print()

print("First 5 rotated points:")
print(rotated_data[:5])
```

```python
# Visualize original and rotated data in 3D

import matplotlib.pyplot as plt

# Create 3D figure
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Original data
ax.scatter(X[:,0], X[:,1], X[:,2])
ax.set_title("Original Dataset")

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

plt.show()
```

```python
# Plot rotated dataset

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.scatter(rotated_data[:,0], rotated_data[:,1], rotated_data[:,2])

ax.set_title("Rotated Dataset (45° around Z-axis)")

ax.set_xlabel("X'")
ax.set_ylabel("Y'")
ax.set_zlabel("Z")

plt.show()
```

```python
# Compare original and rotated data

fig = plt.figure()

ax1 = fig.add_subplot(121, projection='3d')
ax1.scatter(X[:,0], X[:,1], X[:,2])
ax1.set_title("Original Data")

ax2 = fig.add_subplot(122, projection='3d')
ax2.scatter(rotated_data[:,0], rotated_data[:,1], rotated_data[:,2])
ax2.set_title("Rotated Data (45° Z-axis)")

for ax in [ax1, ax2]:
    ax.set_xlabel("X")
    ax.set_ylabel("Y")
    ax.set_zlabel("Z")

plt.show()
```

```python
# Example: rotate an MNIST image

from sklearn.datasets import fetch_openml
import matplotlib.pyplot as plt
import numpy as np
from scipy.ndimage import rotate

# Load MNIST dataset
mnist = fetch_openml('mnist_784', version=1)

# Take one image
image = mnist.data.iloc[0].values.reshape(28,28)

# Rotate image
rotated_image = rotate(image, 45, reshape=False)

# Plot
plt.figure(figsize=(8,4))

plt.subplot(1,2,1)
plt.imshow(image)
plt.title("Original Image")

plt.subplot(1,2,2)
plt.imshow(rotated_image)
plt.title("Rotated Image")

plt.show()
```
