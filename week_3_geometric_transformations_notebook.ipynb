## PART 1: Spatial Dataset Example (City Coordinates)

### Step 2: Create Sample City Dataset

````python
cities = pd.DataFrame({
    'city':["CityA","CityB","CityC","CityD","CityE"],
    'x':[1,4,2,6,5],
    'y':[2,3,5,1,4]
})
cities
````

*italicized text*### Step 3: Plot Original City Coordinates

````python
plt.figure(figsize=(6,6))
plt.scatter(cities['x'], cities['y'])
for i,row in cities.iterrows():
    plt.text(row['x']+0.1,row['y']+0.1,row['city'])
plt.title('Original City Coordinates')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid()
plt.show()
````

### Step 4: Translation Transformation

````python
tx = 3
ty = 2
translated_x = cities['x'] + tx
translated_y = cities['y'] + ty
````



```
# This is formatted as code
```

### Step 5: Plot Translation Result

````python
plt.figure(figsize=(6,6))
plt.scatter(cities['x'], cities['y'], label='Original')
plt.scatter(translated_x, translated_y, label='Translated')
plt.legend()
plt.title('Translation Transformation')
plt.grid()
plt.show()
````

````python
plt.figure(figsize=(6,6))

# Plot original shape with lines
plt.plot(cities['x'], cities['y'], 'bo-', label='Original Shape')

# Plot translated shape with lines
plt.plot(translated_x, translated_y, 'ro-', label='Translated Shape')

# Draw arrows showing translation direction
for i in range(len(cities)):
    plt.arrow(
        cities['x'][i], cities['y'][i],
        translated_x[i] - cities['x'][i],
        translated_y[i] - cities['y'][i],
        head_width=0.15,
        length_includes_head=True,
        color='gray'
    )

# Labels
plt.xlabel("X")
plt.ylabel("Y")

plt.title("Translation Transformation (Shape + Direction)")
plt.legend()
plt.grid()

plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt

# Translation values
tx = 3
ty = 2

# Convert coordinates to matrix form
points = np.column_stack((cities['x'], cities['y']))

# Apply translation
translated_x = cities['x'] + tx
translated_y = cities['y'] + ty

translated_points = np.column_stack((translated_x, translated_y))

# Labels for vertices
labels = ['A','B','C','D','E']

plt.figure(figsize=(6,6))

# Plot original shape
plt.plot(cities['x'], cities['y'], 'bo-', label='Original Shape')

# Plot translated shape
plt.plot(translated_x, translated_y, 'ro-', label='Translated Shape')

# Label original points
for i, label in enumerate(labels):
    plt.text(
        cities['x'][i] + 0.1,
        cities['y'][i] + 0.1,
        label,
        color='blue'
    )

# Label translated points
for i, label in enumerate(labels):
    plt.text(
        translated_x[i] + 0.1,
        translated_y[i] + 0.1,
        label + "'",
        color='red'
    )

# Draw arrows showing movement
for i in range(len(points)):
    plt.arrow(
        points[i,0],
        points[i,1],
        translated_points[i,0] - points[i,0],
        translated_points[i,1] - points[i,1],
        head_width=0.15,
        length_includes_head=True,
        color='gray'
    )

# Plot origin reference
plt.scatter(0,0,color='black')
plt.text(0.1,0.1,'Origin')

plt.xlabel("X")
plt.ylabel("Y")

plt.title("Translation Transformation (tx = 3, ty = 2)")
plt.legend()
plt.grid()

plt.show()
````

### Step 6: Scaling Transformation

````python
Sx = 2
Sy = 0.5
scaled_x = cities['x'] * Sx
scaled_y = cities['y'] * Sy
````

### Step 7 – Plot Scaling Result

````python
plt.figure(figsize=(6,6))
plt.scatter(cities['x'], cities['y'], label='Original')
plt.scatter(scaled_x, scaled_y, label='Scaled')
plt.legend()
plt.title('Scaling Transformation')
plt.grid()
plt.show()
````

````python
plt.figure(figsize=(6,6))

# Plot original shape
plt.plot(cities['x'], cities['y'], 'bo-', label='Original Shape')

# Plot scaled shape
plt.plot(scaled_x, scaled_y, 'ro-', label='Scaled Shape')

# Draw arrows showing scaling movement
for i in range(len(cities)):
    plt.arrow(
        cities['x'][i], cities['y'][i],
        scaled_x[i] - cities['x'][i],
        scaled_y[i] - cities['y'][i],
        head_width=0.15,
        length_includes_head=True,
        color='gray'
    )

plt.xlabel("X")
plt.ylabel("Y")

plt.title("Scaling Transformation (Shape + Direction)")
plt.legend()
plt.grid()

plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt

# Define rotation angle
theta = np.radians(45)

# Rotation matrix
R = np.array([
    [np.cos(theta), -np.sin(theta)],
    [np.sin(theta),  np.cos(theta)]
])

# Convert city coordinates to matrix form
points = np.column_stack((cities['x'], cities['y']))

# Rotate points
rotated = points @ R.T

# Extract rotated coordinates
rotated_x = rotated[:,0]
rotated_y = rotated[:,1]

# Labels for points
labels = ['A','B','C','D','E']

plt.figure(figsize=(6,6))

# Plot original shape
plt.plot(cities['x'], cities['y'], 'bo-', label='Original Shape')

# Plot rotated shape
plt.plot(rotated_x, rotated_y, 'ro-', label='Rotated Shape')

# Label original points
for i, label in enumerate(labels):
    plt.text(
        cities['x'][i] + 0.1,
        cities['y'][i] + 0.1,
        label,
        color='blue'
    )

# Label rotated points
for i, label in enumerate(labels):
    plt.text(
        rotated_x[i] + 0.1,
        rotated_y[i] + 0.1,
        label + "'",
        color='red'
    )

# Draw arrows showing transformation
for i in range(len(points)):
    plt.arrow(
        points[i,0],
        points[i,1],
        rotated_x[i] - points[i,0],
        rotated_y[i] - points[i,1],
        head_width=0.15,
        length_includes_head=True,
        color='gray'
    )

# Plot origin
plt.scatter(0,0,color='black')
plt.text(0.1,0.1,'Origin')

plt.xlabel("X")
plt.ylabel("Y")

plt.title("Rotation Transformation (45° about Origin)")
plt.legend()
plt.grid()

plt.show()
````

### Step 13: Define Trajectory Points

````python
trajectory = np.array([
[0,0],
[1,1],
[2,2],
[3,2],
[4,3],
[5,5]
])
````

### Step 14: Plot Original Trajectory

````python
plt.figure(figsize=(6,5))
plt.plot(trajectory[:,0], trajectory[:,1], marker='o')
plt.title('Original Trajectory')
plt.grid()
plt.show()
````

### Step 15: Reflection Transformation

````python
reflect = np.array([
[1,0],
[0,-1]
])
````

### Step 16: Apply Reflection

````python
reflected = trajectory @ reflect.T
````

### Step 17: Plot Reflection Result

````python
plt.figure(figsize=(6,5))
plt.plot(trajectory[:,0], trajectory[:,1], marker='o', label='Original')
plt.plot(reflected[:,0], reflected[:,1], marker='o', label='Reflected')
plt.legend()
plt.title('Reflection Transformation')
plt.grid()
plt.show()
````
